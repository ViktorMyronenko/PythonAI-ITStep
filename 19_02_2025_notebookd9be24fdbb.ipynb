```python
import pandas as pd
import lightgbm as lgbm

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder

df = pd.read_csv('/kaggle/input/ames-housing-dataset/AmesHousing.csv')
df.drop(columns='Order', inplace=True)
df.head()
```

```python
y = df['SalePrice']
X = df.drop(columns='SalePrice')

from sklearn.model_selection import train_test_split
 
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    train_size=0.8,
                                                    random_state=42)
```

```python
num_columns = X.select_dtypes(include="number").columns
cat_columns = X.select_dtypes(include="object").columns
```

```python
# числові ознаки
num_transformer = Pipeline(
    steps=[("imputer", SimpleImputer(strategy="median"))]
    )
 
# категоріальні ознаки
cat_transformer = Pipeline(
    steps=[("imputer", SimpleImputer(strategy="most_frequent")),
           ("encoder", OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
           ])
 
# об'єднання двох pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_transformer, num_columns),
        ("cat", cat_transformer, cat_columns),
    ]
)
 
```

```python
preprocessor.set_output(transform='pandas')
preprocessor.fit(X_train)
X_train_prep = preprocessor.transform(X_train)
X_test_prep = preprocessor.transform(X_test)
X_test_prep.head()
```

```python
model = lgbm.LGBMRegressor(objective='rmse')
```

```python
model.fit(X_train_prep, y_train,
         eval_set=[(X_train_prep, y_train),(X_test_prep, y_test)],
         eval_names=["X_train", 'X_test'],
         eval_metric=['mse', 'mae'])
```

```python
lgbm.plot_importance(model, max_num_features=20)
```

```python
lgbm.plot_metric(model, metric='l2')
```
